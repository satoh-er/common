# 永田が提案する放射熱伝達率を計算する

## 文献
永田明寛：室内長波放射交換の簡易モデリング、日本建築学会大会学術講演梗概集、pp. 25-26、2016.8 

## 特徴
部位の面積と放射率から、相反則、自己形態係数０を満たす放射熱伝達率を計算する

## 概要
室内各面間の放射伝熱を、空間内の微小球と当該面との放射伝熱で表す。<br>
この時の、各面と微小球の放射熱伝達率と微小球の温度（空間の平均放射温度）を計算する。

## ライブラリ
numpy
scipy.optimize

### 放射熱伝達率
放射熱伝達率は、微小球から面を見た形態係数と放射率から次の式で求められる。<br>

$$
h_r = \frac{\varepsilon}{1 - \varepsilon \cdot f_{mrt}} \cdot 4 \cdot \sigma \cdot (MRT + 273.15)^3
$$

| 記号 | 説明 |
|---|---|
| $h_r$ | 放射熱伝達率 $\mathrm{[W/(m^{2}\!\cdot\!K)]}$ |
| $\varepsilon$ | 放射率 $[-]$ |
| $f_{mrt}$ | 微小球から面を見た形態係数 $[-]$ |
| $\sigma$ | ステファン＝ボルツマン定数 |
| $MRT$ | 平均放射温度 $[^\circ \mathrm{C}]$ |


---

### 微小球から面を見た形態係数
微小球から面を見た形態係数は次式で与えられる。<br>

$$
f_{mrt} = 0.5 \cdot \left( 1 - \sqrt{1 - 4 \cdot \frac{a}{\bar{f}}} \right)
$$

$$
0.5 \cdot \sum_{k=1}^{N} \left( 1 - \sqrt{1 - 4 \cdot \frac{a}{\bar{f}}} \right) = 1
$$

| 記号 | 説明 |
|---|---|
| $f_{mrt}$ | 微小球から面を見た形態係数 $[-]$ |
| $a$ | 空間の全面積に対する当該部位の面積の比 $[-]$ |
| $\bar{f}$ | 形態係数の総和が1となるように調整された代表値 |
| $k$ | 空間の面番号 |
| $N$ | 空間の面の総数 |



### 例題
間口5m、奥行6m、高さ3mの空間の放射熱伝達率を計算する。<br>
面積は以下の通り。<br>
床・天井：30m2<br>
壁：18m2×2面、15m2×2面 <br>
各面の放射率は一律0.9とする。

```python
# 面積を有する配列の作成
area = np.array([30.0, 30.0, 18.0, 18.0, 15.0, 15.0])

# 放射率を有する配列の作成
emissivity = np.full(len(area), 0.9)
```

```python
# 放射熱伝達率の計算
h_r = radiative.calc_radiative_heat_transfer_coeff(area=area, emissivity=emissivity, mrt=20)

print(h_r)
```

```python
# 微小球の放射温度計算における各面の重み
weights = radiative.calc_weights_for_mrt(h_r=h_r, area=area)

print(weights)
print(np.sum(weights))
```
